```python
occupancy_file = data_folder + "/5occupancySensor_MayToDec2024_9MRows.csv"

# Only load the first 1,000 rows for investigation
occupancy_sample = pd.read_csv(occupancy_file, nrows=1000)

print("Sample size:", occupancy_sample.shape)

print("\nColumn names:")
print(occupancy_sample.columns.tolist())

print("\nData types:")
print(occupancy_sample.dtypes)

occupancy_sample.head()
```

```python
print("Occupancy statuses:")
print(occupancy_sample["occupancystatus"].value_counts())

print("\nPrevious occupancy statuses:")
print(occupancy_sample["previousoccupancystatus"].value_counts())

print("\nHeadcount values:")
print(occupancy_sample["headcount"].value_counts().sort_index())

print("\nUnique device IDs:")
print(occupancy_sample["deviceid"].nunique())

print("\nUnique floor space IDs:")
print(occupancy_sample["floorspaceid"].nunique())

print("\nDate range in sample:")
print(occupancy_sample["collecteddate"].min())
print(occupancy_sample["collecteddate"].max())
```

```python
# Investigate the full occupancy dataset in manageable chunks

use_columns = [
    "deviceid",
    "floorspaceid",
    "occupancystatus",
    "headcount",
    "collecteddate"
]

total_rows = 0
device_ids = set()
floorspace_ids = set()
min_date = None
max_date = None

for chunk in pd.read_csv(
    occupancy_file,
    usecols=use_columns,
    chunksize=500000
):
    total_rows += len(chunk)

    device_ids.update(chunk["deviceid"].dropna().unique())
    floorspace_ids.update(chunk["floorspaceid"].dropna().unique())

    dates = pd.to_datetime(chunk["collecteddate"], errors="coerce", utc=True)

    chunk_min = dates.min()
    chunk_max = dates.max()

    if min_date is None or chunk_min < min_date:
        min_date = chunk_min

    if max_date is None or chunk_max > max_date:
        max_date = chunk_max

print("Total rows:", total_rows)
print("Unique devices:", len(device_ids))
print("Unique floor spaces:", len(floorspace_ids))
print("Earliest date:", min_date)
print("Latest date:", max_date)
```

```python
# Count records for each floor space

floorspace_counts = {}

for chunk in pd.read_csv(
    occupancy_file,
    usecols=["floorspaceid"],
    chunksize=500000
):
    counts = chunk["floorspaceid"].value_counts()

    for space_id, count in counts.items():
        floorspace_counts[space_id] = floorspace_counts.get(space_id, 0) + count

floorspace_counts
```

```python
# Count environmental records for each sensor

print("Environmental records per sensor:")
print(env["sensorid"].value_counts())

print("\nDate range for environmental data:")
env_dates = pd.to_datetime(env["createdate"], errors="coerce", utc=True)

print("Earliest:", env_dates.min())
print("Latest:", env_dates.max())
```

```python
# Search the location spreadsheet for the 5 occupancy floor-space IDs

floor_ids = list(floorspace_counts.keys())

for floor_id in floor_ids:
    matches = locations[
        locations.astype(str).apply(
            lambda row: row.str.contains(floor_id, case=False, na=False).any(),
            axis=1
        )
    ]

    print("\nFloor space ID:", floor_id)

    if len(matches) > 0:
        print(matches)
    else:
        print("No match found")
```

```python
import json
import pandas as pd

def extract_measurements(json_string):
    data = json.loads(json_string)
    measurements = {}

    for item in data:
        name = item.get("variable", {}).get("name")

        values = item.get("values", [])

        if name and values:
            first_value = values[0]

            # Some rows do not contain a normal "value" field
            if isinstance(first_value, dict) and "value" in first_value:
                measurements[name] = first_value["value"]

    return pd.Series(measurements)

env_measurements = env["jsondata"].apply(extract_measurements)

env_clean = pd.concat([
    env[["sensorid", "createdate"]].reset_index(drop=True),
    env_measurements.reset_index(drop=True)
], axis=1)

print("Clean dataset size:", env_clean.shape)

print("\nColumns:")
print(env_clean.columns.tolist())

env_clean.head()
```

```python
# Check missing values in the cleaned environmental data

print("Missing values:")
print(env_clean.isnull().sum())

print("\nData types:")
print(env_clean.dtypes)
```

```python
# Summary statistics for the main environmental variables

main_variables = [
    "Carbon dioxide",
    "Temperature",
    "Humidity"
]

env_clean[main_variables].describe()
```

```python
# Compare the main environmental variables between sensors

sensor_summary = env_clean.groupby("sensorid")[
    ["Carbon dioxide", "Temperature", "Humidity"]
].mean().round(2)

sensor_summary
```

```python
# Convert environmental dates to proper datetime format
env_clean["createdate"] = pd.to_datetime(
    env_clean["createdate"],
    errors="coerce",
    utc=True
)

# Check the date range and number of records for each sensor
sensor_dates = env_clean.groupby("sensorid").agg(
    first_reading=("createdate", "min"),
    last_reading=("createdate", "max"),
    number_of_readings=("createdate", "count")
)

sensor_dates
```

```python
processed_file = data_folder + "/env_clean.csv"

env_clean.to_csv(processed_file, index=False)

print("Saved successfully:")
print(processed_file)
```
